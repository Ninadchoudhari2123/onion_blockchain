```python
class Block:

    def __init__(self, index, transactions, previous_hash):

        self.index = index

        self.timestamp = str(datetime.now())

        self.transactions = transactions

        self.previous_hash = previous_hash

        self.hash = self.calculate_hash()

    def calculate_hash(self):

        block_data = {
            "index": self.index,
            "timestamp": self.timestamp,
            "transactions": self.transactions,
            "previous_hash": self.previous_hash
        }

        block_string = json.dumps(
            block_data,
            sort_keys=True
        )

        return hashlib.sha256(
            block_string.encode()
        ).hexdigest()
```

```python
class Blockchain:

    def __init__(self):

        self.chain = []

        self.create_genesis_block()

    def create_genesis_block(self):

        genesis_block = Block(
            0,
            [{"event": "Genesis Block"}],
            "0"
        )

        self.chain.append(genesis_block)

    def add_block(self, transactions):

        previous_block = self.chain[-1]

        new_block = Block(
            len(self.chain),
            transactions,
            previous_block.hash
        )

        self.chain.append(new_block)

        return new_block

    def validate_chain(self):

        for i in range(1, len(self.chain)):

            current_block = self.chain[i]

            previous_block = self.chain[i - 1]

            if current_block.hash != current_block.calculate_hash():
                return False

            if current_block.previous_hash != previous_block.hash:
                return False

        return True
```

```python
import json
import base64

from datetime import datetime

from cryptography.hazmat.primitives.asymmetric.ed25519 import Ed25519PrivateKey

from cryptography.exceptions import InvalidSignature

print("Security module initialized")
```

```python
actors = {}

print("Participant storage created")
```

```python
def create_actor(actor_id, role):

    private_key = Ed25519PrivateKey.generate()

    public_key = private_key.public_key()

    actors[actor_id] = {
        "role": role,
        "private_key": private_key,
        "public_key": public_key
    }

    print(actor_id, "registered successfully")
```

```python
create_actor("farmer_001", "FARMER")

create_actor("center_001", "COLLECTION_CENTER")

create_actor("transporter_001", "TRANSPORTER")

create_actor("wholesaler_001", "WHOLESALER")

create_actor("retailer_001", "RETAILER")
```

```python
permissions = {

    "FARMER": [
        "BATCH_CREATED",
        "HARVESTED"
    ],

    "COLLECTION_CENTER": [
        "RECEIVED",
        "QUALITY_INSPECTED",
        "DISPATCHED"
    ],

    "TRANSPORTER": [
        "PICKED_UP",
        "DELIVERED"
    ],

    "WHOLESALER": [
        "RECEIVED",
        "OWNERSHIP_TRANSFER",
        "DISPATCHED"
    ],

    "RETAILER": [
        "RECEIVED"
    ]
}

print("Participant permissions configured")
```

```python
def create_transaction(
    actor_id,
    batch_id,
    event_type,
    from_state,
    to_state,
    quantity_kg
):

    transaction = {
        "transaction_id": "TXN-" + str(datetime.now().timestamp()),
        "batch_id": batch_id,
        "actor_id": actor_id,
        "event_type": event_type,
        "from_state": from_state,
        "to_state": to_state,
        "quantity_kg": quantity_kg,
        "timestamp": datetime.now().isoformat(),
        "record_origin": "DEMO"
    }

    data = json.dumps(
        transaction,
        sort_keys=True,
        separators=(",", ":")
    ).encode()

    signature = actors[actor_id]["private_key"].sign(data)

    transaction["signature"] = base64.b64encode(
        signature
    ).decode()

    return transaction
```

```python
def verify_signature(transaction):

    actor_id = transaction.get("actor_id")

    if actor_id not in actors:
        return False

    new_transaction = transaction.copy()

    signature = new_transaction.pop("signature", None)

    if signature is None:
        return False

    data = json.dumps(
        new_transaction,
        sort_keys=True,
        separators=(",", ":")
    ).encode()

    try:

        actors[actor_id]["public_key"].verify(
            base64.b64decode(signature),
            data
        )

        return True

    except (InvalidSignature, ValueError, TypeError):

        return False
```

```python
def validate_transaction(transaction):

    actor_id = transaction.get("actor_id")

    if actor_id not in actors:

        return False, "Participant not registered"

    role = actors[actor_id]["role"]

    event_type = transaction.get("event_type")

    if event_type not in permissions.get(role, []):

        return False, "Permission denied"

    if not verify_signature(transaction):

        return False, "Invalid digital signature"

    quantity = transaction.get("quantity_kg")

    if not isinstance(quantity, (int, float)) or quantity <= 0:

        return False, "Invalid quantity"

    return True, "Transaction verified successfully"
```

```python
import uuid


def create_supply_transaction(
    actor_id,
    batch_id,
    event_type,
    quantity_kg,
    from_state="",
    to_state="",
    to_actor="",
    details=None
):

    transaction = {

        "transaction_id": "TXN-" + str(uuid.uuid4()),

        "batch_id": batch_id,

        "actor_id": actor_id,

        "event_type": event_type,

        "from_actor": actor_id,

        "to_actor": to_actor,

        "from_state": from_state,

        "to_state": to_state,

        "quantity_kg": quantity_kg,

        "details": details or {},

        "timestamp": datetime.now().isoformat(),

        "record_origin": "DEMO"
    }

    data = json.dumps(
        transaction,
        sort_keys=True,
        separators=(",", ":")
    ).encode()

    signature = actors[actor_id]["private_key"].sign(data)

    transaction["signature"] = base64.b64encode(
        signature
    ).decode()

    return transaction
```

```python
batch_data = {}

batch_history = {}

for block in agri_chain.chain:

    for tx in block.transactions:

        batch_id = tx.get("batch_id")

        event_type = tx.get("event_type")

        if batch_id:

            batch_history.setdefault(batch_id, []).append(tx)

        if event_type == "BATCH_CREATED":

            batch_data[batch_id] = {

                "quantity_kg": tx["quantity_kg"],

                "current_holder": tx["actor_id"],

                "current_state": tx["to_state"],

                "status": "CREATED"
            }

        elif event_type == "OWNERSHIP_TRANSFER":

            if batch_id in batch_data:

                batch_data[batch_id]["current_holder"] = tx["to_actor"]

                batch_data[batch_id]["current_state"] = tx["to_state"]

        elif event_type == "DELIVERED":

            if batch_id in batch_data:

                batch_data[batch_id]["current_state"] = tx["to_state"]


print("Batch storage initialized")

print("Existing batches:", len(batch_data))
```

```python
def validate_supply_event(tx):

    status, message = validate_transaction(tx)

    if not status:

        return False, message

    if not agri_chain.validate_chain():

        return False, "Blockchain integrity failed"

    batch_id = tx["batch_id"]

    actor_id = tx["actor_id"]

    event_type = tx["event_type"]

    quantity = tx["quantity_kg"]

    role = actors[actor_id]["role"]

    if event_type == "BATCH_CREATED":

        if batch_id in batch_data:

            return False, "Batch already exists"

        if role != "FARMER":

            return False, "Only farmers can create batches"

        if not tx["to_state"]:

            return False, "Farm location is required"

        return True, "New batch is valid"

    if batch_id not in batch_data:

        return False, "Batch not found"

    batch = batch_data[batch_id]

    if actor_id != batch["current_holder"]:

        return False, "Participant is not the current batch holder"

    if quantity != batch["quantity_kg"]:

        return False, "Batch quantity mismatch"

    if tx["from_state"] != batch["current_state"]:

        return False, "Incorrect source location"

    if event_type == "OWNERSHIP_TRANSFER":

        to_actor = tx["to_actor"]

        if to_actor not in actors:

            return False, "Recipient is not registered"

        if to_actor == actor_id:

            return False, "Cannot transfer batch to yourself"

        if not tx["to_state"]:

            return False, "Destination state is required"

    elif event_type == "DELIVERED":

        if not tx["to_state"]:

            return False, "Delivery location is required"

    elif event_type not in [
        "HARVESTED",
        "QUALITY_INSPECTED",
        "PICKED_UP",
        "RECEIVED"
    ]:

        return False, "Unsupported supply-chain event"

    return True, "Supply-chain transaction verified"
```

```python
def submit_supply_event(tx):

    status, message = validate_supply_event(tx)

    if not status:

        return False, message

    agri_chain.add_block([tx])

    batch_id = tx["batch_id"]

    event_type = tx["event_type"]

    if event_type == "BATCH_CREATED":

        batch_data[batch_id] = {

            "quantity_kg": tx["quantity_kg"],

            "current_holder": tx["actor_id"],

            "current_state": tx["to_state"],

            "status": "CREATED"
        }

    elif event_type == "OWNERSHIP_TRANSFER":

        batch_data[batch_id]["current_holder"] = tx["to_actor"]

        batch_data[batch_id]["current_state"] = tx["to_state"]

    elif event_type == "DELIVERED":

        batch_data[batch_id]["current_state"] = tx["to_state"]

    if event_type == "RECEIVED":

        batch_data[batch_id]["status"] = "RECEIVED"

    batch_history.setdefault(batch_id, []).append(tx)

    return True, "Transaction added successfully"
```

```python
def split_batch(actor_id, parent_id, child_batches):

    if parent_id not in batch_data:
        return False, "Parent batch not found"

    parent = batch_data[parent_id]

    if parent["status"] in ["SPLIT", "MERGED"]:
        return False, "Parent batch is not active"

    if parent["current_holder"] != actor_id:
        return False, "Only current holder can split batch"

    if actors[actor_id]["role"] not in [
        "COLLECTION_CENTER",
        "WHOLESALER"
    ]:
        return False, "Permission denied"

    if not isinstance(child_batches, list) or len(child_batches) < 2:
        return False, "At least two child batches required"

    child_ids = []

    total_quantity = 0

    for child in child_batches:

        child_id = child["batch_id"]

        quantity = child["quantity_kg"]

        if child_id in batch_data or child_id in child_ids:
            return False, "Duplicate child batch ID"

        if not isinstance(quantity, (int, float)) or quantity <= 0:
            return False, "Invalid child quantity"

        child_ids.append(child_id)

        total_quantity += quantity

    if total_quantity != parent["quantity_kg"]:
        return False, "Child quantities do not match parent quantity"

    location = parent["current_state"]

    tx = create_supply_transaction(

        actor_id=actor_id,

        batch_id=parent_id,

        event_type="BATCH_SPLIT",

        quantity_kg=parent["quantity_kg"],

        from_state=location,

        to_state=location,

        to_actor=actor_id,

        details={
            "child_batches": child_batches
        }
    )

    status, message = validate_transaction(tx)

    if not status:
        return False, message

    if not agri_chain.validate_chain():
        return False, "Blockchain integrity failed"

    agri_chain.add_block([tx])

    batch_data[parent_id]["status"] = "SPLIT"

    batch_lineage.setdefault(
        parent_id,
        {"parents": [], "children": []}
    )

    batch_lineage[parent_id]["children"] = child_ids

    batch_history.setdefault(parent_id, []).append(tx)

    for child in child_batches:

        child_id = child["batch_id"]

        batch_data[child_id] = {
            "quantity_kg": child["quantity_kg"],
            "current_holder": actor_id,
            "current_state": location,
            "status": "ACTIVE"
        }

        batch_lineage[child_id] = {
            "parents": [parent_id],
            "children": []
        }

        batch_history[child_id] = []

    return True, "Batch split successfully"
```

```python
def merge_batches(actor_id, parent_ids, new_batch_id):

    if new_batch_id in batch_data:
        return False, "New batch ID already exists"

    if not isinstance(parent_ids, list) or len(parent_ids) < 2:
        return False, "At least two parent batches required"

    if len(parent_ids) != len(set(parent_ids)):
        return False, "Duplicate parent batch ID"

    total_quantity = 0

    location = ""

    for parent_id in parent_ids:

        if parent_id not in batch_data:
            return False, "Parent batch not found"

        parent = batch_data[parent_id]

        if parent["status"] in ["SPLIT", "MERGED"]:
            return False, "Parent batch is not active"

        if parent["current_holder"] != actor_id:
            return False, "Participant is not the current holder"

        if not location:
            location = parent["current_state"]

        elif parent["current_state"] != location:
            return False, "Batches must be at the same location"

        total_quantity += parent["quantity_kg"]

    if actors[actor_id]["role"] not in [
        "COLLECTION_CENTER",
        "WHOLESALER"
    ]:
        return False, "Permission denied"

    tx = create_supply_transaction(

        actor_id=actor_id,

        batch_id=new_batch_id,

        event_type="BATCH_MERGED",

        quantity_kg=total_quantity,

        from_state=location,

        to_state=location,

        to_actor=actor_id,

        details={
            "parent_batches": parent_ids
        }
    )

    status, message = validate_transaction(tx)

    if not status:
        return False, message

    if not agri_chain.validate_chain():
        return False, "Blockchain integrity failed"

    agri_chain.add_block([tx])

    for parent_id in parent_ids:

        batch_data[parent_id]["status"] = "MERGED"

        batch_data[parent_id]["merged_into"] = new_batch_id

        batch_history.setdefault(parent_id, []).append(tx)

        batch_lineage.setdefault(
            parent_id,
            {"parents": [], "children": []}
        )

        if new_batch_id not in batch_lineage[parent_id]["children"]:
            batch_lineage[parent_id]["children"].append(new_batch_id)

    batch_data[new_batch_id] = {

        "quantity_kg": total_quantity,

        "current_holder": actor_id,

        "current_state": location,

        "status": "ACTIVE"
    }

    batch_lineage[new_batch_id] = {

        "parents": parent_ids,

        "children": []
    }

    batch_history[new_batch_id] = [tx]

    return True, "Batches merged successfully"
```

```python
def get_batch_history(batch_id):

    if batch_id not in batch_data:
        return []

    events = []

    parents = batch_lineage.get(
        batch_id,
        {"parents": [], "children": []}
    )["parents"]

    for parent_id in parents:

        events.extend(get_batch_history(parent_id))

    events.extend(batch_history.get(batch_id, []))

    unique_events = {}

    for tx in events:

        unique_events[tx["transaction_id"]] = tx

    result = list(unique_events.values())

    result.sort(key=lambda x: x["timestamp"])

    return result
```

```python
def save_actors():

    for actor_id, actor in actors.items():

        public_key = actor["public_key"].public_bytes(
            encoding=serialization.Encoding.PEM,
            format=serialization.PublicFormat.SubjectPublicKeyInfo
        ).decode()

        conn.execute("""
            INSERT INTO actors (
                actor_id,
                role,
                public_key,
                created_at
            )
            VALUES (?, ?, ?, ?)

            ON CONFLICT(actor_id)
            DO UPDATE SET
                role = excluded.role,
                public_key = excluded.public_key

        """, (
            actor_id,
            actor["role"],
            public_key,
            datetime.now().isoformat()
        ))

    conn.commit()

    print("Participants saved successfully")
```

```python
def save_blockchain():

    for block in agri_chain.chain:

        block_data = json.dumps(
            block.transactions,
            sort_keys=True,
            default=str
        )

        conn.execute("""
            INSERT INTO blocks (
                block_index,
                timestamp,
                transactions_json,
                previous_hash,
                current_hash
            )
            VALUES (?, ?, ?, ?, ?)

            ON CONFLICT(block_index)
            DO UPDATE SET
                timestamp = excluded.timestamp,
                transactions_json = excluded.transactions_json,
                previous_hash = excluded.previous_hash,
                current_hash = excluded.current_hash

        """, (
            block.index,
            block.timestamp,
            block_data,
            block.previous_hash,
            block.hash
        ))

        for tx in block.transactions:

            tx_id = tx.get("transaction_id")

            if not tx_id:
                continue

            tx_json = json.dumps(
                tx,
                sort_keys=True,
                default=str
            )

            conn.execute("""
                INSERT OR REPLACE INTO chain_transactions (

                    transaction_id,
                    block_index,
                    batch_id,
                    actor_id,
                    event_type,
                    from_actor,
                    to_actor,
                    from_state,
                    to_state,
                    quantity_kg,
                    record_origin,
                    timestamp,
                    transaction_json

                )

                VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)

            """, (

                str(tx_id),
                block.index,
                tx.get("batch_id"),
                tx.get("actor_id"),
                tx.get("event_type"),
                tx.get("from_actor"),
                tx.get("to_actor"),
                tx.get("from_state"),
                tx.get("to_state"),
                tx.get("quantity_kg"),
                tx.get("record_origin"),
                tx.get("timestamp"),
                tx_json

            ))

    conn.commit()

    print("Blockchain saved successfully")
```

```python
def save_batch_lineage():

    conn.execute("DELETE FROM batch_lineage")

    saved_links = set()

    for batch_id, info in batch_lineage.items():

        for parent_id in info.get("parents", []):

            saved_links.add((parent_id, batch_id))

        for child_id in info.get("children", []):

            saved_links.add((batch_id, child_id))

    for parent_id, child_id in saved_links:

        conn.execute("""
            INSERT OR IGNORE INTO batch_lineage (
                parent_batch_id,
                child_batch_id
            )

            VALUES (?, ?)

        """, (parent_id, child_id))

    conn.commit()

    print("Batch lineage saved successfully")
```

```python
def save_market_data(market_df):

    columns = [
        "state",
        "district",
        "market",
        "commodity",
        "variety",
        "grade",
        "arrival_date",
        "min_price",
        "max_price",
        "modal_price"
    ]

    for col in columns:

        if col not in market_df.columns:
            raise ValueError("Missing column: " + col)

    count = 0

    for row in market_df[columns].to_dict("records"):

        row_data = {}

        for col in columns:

            value = row[col]

            if pd.isna(value):
                value = None

            elif col in [
                "min_price",
                "max_price",
                "modal_price"
            ]:
                value = float(value)

            else:
                value = str(value)

            row_data[col] = value

        row_text = json.dumps(
            row_data,
            sort_keys=True
        )

        row_hash = hashlib.sha256(
            row_text.encode()
        ).hexdigest()

        conn.execute("""
            INSERT OR IGNORE INTO market_prices (

                source_row_hash,
                state,
                district,
                market,
                commodity,
                variety,
                grade,
                arrival_date,
                min_price,
                max_price,
                modal_price

            )

            VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)

        """, (

            row_hash,
            row_data["state"],
            row_data["district"],
            row_data["market"],
            row_data["commodity"],
            row_data["variety"],
            row_data["grade"],
            row_data["arrival_date"],
            row_data["min_price"],
            row_data["max_price"],
            row_data["modal_price"]

        ))

        count += 1

    conn.commit()

    total = conn.execute(
        "SELECT COUNT(*) FROM market_prices"
    ).fetchone()[0]

    print("Rows processed:", count)

    print("Unique records stored:", total)
```

```python

import streamlit as st
import pandas as pd
import sqlite3

from pathlib import Path

# page configuration
st.set_page_config(
    page_title="AgriTrace",
    page_icon="🌾",
    layout="wide"
)

# project paths
base_dir = Path(__file__).resolve().parent

db_path = base_dir / "database" / "agritrace.db"

market_file = (
    base_dir /
    "data" /
    "official" /
    "processed" /
    "all_india_onion_data.csv"
)

# load database
def load_data(table):

    if not db_path.exists():
        return pd.DataFrame()

    with sqlite3.connect(db_path) as conn:

        tables = pd.read_sql_query(
            "SELECT name FROM sqlite_master WHERE type='table'",
            conn
        )["name"].tolist()

        if table not in tables:
            return pd.DataFrame()

        return pd.read_sql_query(
            f"SELECT * FROM {table}",
            conn
        )


# load datasets
market_df = load_data("market_prices")
blocks = load_data("blocks")
batches = load_data("batches")
transactions = load_data("chain_transactions")
actors = load_data("actors")


# use csv if market table is empty
if market_df.empty and market_file.exists():

    market_df = pd.read_csv(market_file)


# sidebar
st.sidebar.title("🌾 AgriTrace")

page = st.sidebar.radio(
    "Navigation",
    [
        "Dashboard",
        "Market Analytics",
        "Batch Management",
        "Blockchain Explorer"
    ]
)


# dashboard
if page == "Dashboard":

    st.title("🌾 AgriTrace")
    st.subheader(
        "Blockchain-Based Agricultural Product Traceability"
    )

    col1, col2, col3, col4 = st.columns(4)

    col1.metric("Market Records", len(market_df))

    col2.metric(
        "States",
        market_df["state"].nunique()
        if "state" in market_df.columns else 0
    )

    col3.metric("Blockchain Blocks", len(blocks))

    col4.metric("Onion Batches", len(batches))

    st.divider()

    if not market_df.empty:

        st.subheader("State-wise Onion Market Records")

        state_data = market_df["state"].value_counts().head(10)

        st.bar_chart(state_data)

    st.subheader("Recent Blockchain Transactions")

    if not transactions.empty:
        st.dataframe(
            transactions.tail(10),
            use_container_width=True
        )
    else:
        st.info("No transactions available.")


# market analytics
elif page == "Market Analytics":

    st.title("📊 All-India Onion Market Analytics")

    if market_df.empty:
        st.warning("Market dataset not found.")

    else:

        states = sorted(market_df["state"].dropna().unique())

        selected_state = st.selectbox(
            "Select State",
            ["All India"] + states
        )

        filtered_data = market_df.copy()

        if selected_state != "All India":

            filtered_data = filtered_data[
                filtered_data["state"] == selected_state
            ]

        st.metric("Total Records", len(filtered_data))

        st.metric(
            "Total Markets",
            filtered_data["market"].nunique()
        )

        filtered_data["modal_price"] = pd.to_numeric(
            filtered_data["modal_price"],
            errors="coerce"
        )

        avg_price = filtered_data["modal_price"].mean()

        st.metric(
            "Average Modal Price",
            f"₹{avg_price:,.2f}" if pd.notna(avg_price) else "N/A"
        )

        market_price = (
            filtered_data.groupby("market")["modal_price"]
            .mean()
            .sort_values(ascending=False)
            .head(10)
        )

        st.subheader("Top 10 Markets by Average Modal Price")

        st.bar_chart(market_price)

        st.dataframe(
            filtered_data,
            use_container_width=True
        )


# batch management
elif page == "Batch Management":

    st.title("📦 Onion Batch Management")

    if batches.empty:
        st.info("No batches found in the database.")

    else:

        st.dataframe(
            batches,
            use_container_width=True
        )


# blockchain explorer
elif page == "Blockchain Explorer":

    st.title("⛓️ Blockchain Explorer")

    if blocks.empty:
        st.info("No blockchain blocks found.")

    else:

        for _, block in blocks.iterrows():

            with st.expander(
                f"Block #{block['block_index']}"
            ):

                st.write("Timestamp:", block["timestamp"])

                st.write(
                    "Previous Hash:",
                    block["previous_hash"]
                )

                st.write(
                    "Current Hash:",
                    block["current_hash"]
                )

                st.json(
                    __import__("json").loads(
                        block["transactions_json"]
                    )
                )
```
